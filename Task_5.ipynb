```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading dataset
titanic = df = pd.read_csv('titanic.csv')

# Basic inspection
print("First 5 Rows of Dataset:")
df.head()
```

```python
print("\nDataset Info:")
df.info()

print("\nMissing Values:")
df.isnull().sum()
```

```python
print("\nStatistical Summary:")
df.describe()
```

```python
# Categorical columns value counts
print(df['Survived'].value_counts(),"\n")
print(df['Pclass'].value_counts(),"\n")
print(df['Sex'].value_counts(),"\n")
print(df['Embarked'].value_counts(),"\n")
```

```python
# ---------------- HISTOGRAMS ----------------

# Histogram for Age
plt.figure(figsize=(8,5))
sns.histplot(data=titanic, x='Age', kde=True, color='skyblue')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(True)
plt.show()

# Histogram for Fare
plt.figure(figsize=(8,5))
sns.histplot(data=titanic, x='Fare', kde=True, color='lightgreen')
plt.title('Distribution of Fare')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.grid(True)
plt.show()

# Countplot for Pclass (fixed future warning by adding hue)
plt.figure(figsize=(8,5))
sns.countplot(data=titanic, x='Pclass', hue='Pclass', palette='pastel', legend=False)
plt.title('Passenger Class Distribution')
plt.xlabel('Passenger Class (1=Upper, 2=Middle, 3=Lower)')
plt.ylabel('Count')
plt.grid(True)
plt.show()

# ---------------- BOX PLOTS ----------------

# Boxplot of Age by Survival
plt.figure(figsize=(8,5))
sns.boxplot(data=titanic, x='Survived', y='Age', palette='Set2')
plt.title('Age Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.grid(True)
plt.show()

# Boxplot of Fare by Pclass
plt.figure(figsize=(8,5))
sns.boxplot(data=titanic, x='Pclass', y='Fare', palette='Set3')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.grid(True)
plt.show()

# ---------------- SCATTER PLOTS ----------------

# Scatterplot Age vs Fare colored by Survival
plt.figure(figsize=(8,5))
sns.scatterplot(data=titanic, x='Age', y='Fare', hue='Survived', palette='coolwarm')
plt.title('Age vs Fare (Colored by Survival)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='Survived')
plt.grid(True)
plt.show()
```

Observations for Plots: 

1. Histogram for Age
The age distribution shows a peak between 20–30 years, meaning most passengers were young adults.
Very few elderly passengers (above 60 years) traveled.
There are some missing values (gaps in the histogram).
The distribution is slightly skewed right — more young passengers than old.

2. Histogram for Fare
The majority of passengers paid a fare less than $50.
A few extreme high fares ($200–$500) are visible — likely paid by First Class travelers.
The fare distribution is highly skewed to the right due to these expensive outliers.
This suggests class inequality: most could afford only cheap tickets.

3. Countplot for Passenger Class (Pclass)
Third Class (Pclass 3) passengers formed the largest group.
First Class passengers were the fewest but likely the wealthiest.
This suggests the Titanic catered mostly to lower/middle class travelers.

4. Boxplot of Age by Survival
Children (young age group) had a better chance of survival.
Older passengers had a slightly lower survival probability.
The median age of survivors is lower than that of non-survivors.
Age played an important role in survival decisions (e.g., "women and children first" policy).

5. Boxplot of Fare by Pclass
A clear upward trend: First Class → Second Class → Third Class.
First class fares have a wide range, indicating even within first class, there were expensive and more moderate tickets.
Third class fares are tightly packed — mostly affordable.

6. Scatterplot Age vs Fare (colored by Survival)
Passengers paying higher fares had better survival chances (more survivors in red/blue at high fare areas).
There is no strong direct relation between age and fare.
Few very young passengers with high fares survived — possibly children of wealthy families.
